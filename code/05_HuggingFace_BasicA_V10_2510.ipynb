## Hugging Face 기본 예제

```python
# Hugging Face의 Transformers 라이브러리 사용하기
from transformers import pipeline

# 감정 분석 모델 불러오기 (한 줄로!)
classifier = pipeline("sentiment-analysis")

# 텍스트의 감정 분석하기
result = classifier("이 제품 정말 좋아요!")
print(result)

# 출력 결과: [{'label': 'POSITIVE', 'score': 0.9998}]
# 긍정적 감정이 99.98% 확률로 감지되었습니다
```

## 텍스트 생성

```python
# ============================================================
# 【초급 2】 텍스트 생성 (Text Generation)
# "문장을 시작하면 AI가 이어서 작성해줍니다"
# ============================================================

from transformers import pipeline

# 한국어 텍스트 생성 모델 사용
generator = pipeline("text-generation", model="skt/kogpt2-base-v2")

prompt = "옛날 옛적에"
result = generator(
    prompt,
    max_length=100,  # 한국어는 좀 더 길게 설정
    num_return_sequences=1,
    pad_token_id=50256  # 경고 메시지 방지
)

print(f"\n시작 문장: {prompt}")
print(f"AI가 완성한 문장:\n{result[0]['generated_text']}")
```

## 여러 버전 생성해 보기

```python
# ============================================================
# 추가 예제: 여러 버전 생성해보기
# ============================================================
print("\n" + "="*60)
print("🎨 여러 버전으로 생성해보기")
print("="*60)

prompt2 = "인공지능의 미래는"
results = generator(
    prompt2,
    max_length=80,
    num_return_sequences=3,  # 3가지 버전 생성
    do_sample=True,  # 샘플링 활성화
    temperature=0.8,  # 창의성 조절
    pad_token_id=50256
)

for i, result in enumerate(results, 1):
    print(f"\n[버전 {i}]")
    print(result['generated_text'])
```

## 🌍 텍스트 번역

```python
from transformers import pipeline

# 영어 → 프랑스어
translator = pipeline("translation_en_to_fr")
result = translator("Hello, how are you?")
print(result[0]['translation_text'])

# 영어 → 독일어
de_translator = pipeline("translation_en_to_de")
print(de_translator("Good morning!")[0]['translation_text'])

# 여러 문장 번역
sentences = [
    "I love programming.",
    "The weather is beautiful today.",
    "Let's meet tomorrow."
]
translations = translator(sentences)
for orig, trans in zip(sentences, translations):
    print(f"🇬🇧 {orig}\n🇫🇷 {trans['translation_text']}\n")
```

## 텍스트 요약

```python
from transformers import pipeline

# 요약 파이프라인
summarizer = pipeline("summarization")

# 긴 기사 예제
long_article = """
Artificial intelligence has made remarkable progress in recent years.
Machine learning models can now perform tasks that were once thought
to be exclusively human, such as recognizing images, understanding
speech, and even generating creative content. The technology is being
applied across various industries including healthcare, finance, and
transportation. However, concerns about AI safety and ethics continue
to be important topics of discussion among researchers and policymakers.
"""

# 요약 생성
summary = summarizer(long_article, max_length=50, min_length=25)
print("📄 원문 길이:", len(long_article.split()))
print("📝 요약 길이:", len(summary[0]['summary_text'].split()))
print("\n✨ 요약 결과:")
print(summary[0]['summary_text'])
```
